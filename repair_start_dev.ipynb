```python
import subprocess
from pathlib import Path
from time import sleep
root = Path(r"C:\ProgramData\MySQL\MySQL Server 8.0")
my_ini = root / 'my.ini'
mysqld = root / 'bin' / 'mysqld.exe'
print('mysqld path', mysqld.exists(), mysqld)
print('my.ini path', my_ini.exists(), my_ini)
if mysqld.exists() and my_ini.exists():
    proc = subprocess.Popen([str(mysqld), f"--defaults-file={my_ini}", '--console'], stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    print('started pid', proc.pid)
    sleep(5)
    print('poll', proc.poll())
    if proc.poll() is not None:
        out, err = proc.communicate(timeout=5)
        print('process exited early')
        print('out', out)
        print('err', err)
    else:
        print('process still running; checking ports')
        import socket
        for port in [3306, 3307, 33060]:
            s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            s.settimeout(1)
            try:
                s.connect(('127.0.0.1', port))
                print(port, 'open')
            except Exception as e:
                print(port, 'closed', type(e).__name__, e)
            finally:
                s.close()
        proc.terminate()
        proc.wait(timeout=10)
        print('terminated', proc.returncode)
else:
    print('mysqld or my.ini missing')
```

```python
from pathlib import Path
script = Path(r"C:\Users\Виктор\Downloads\PRO100\PRO100\PRO100v6.43x64\Проекты\shop\start-dev.ps1")
contents = '''# Local development startup script for the shop project
# Run with PowerShell: .\start-dev.ps1

$ErrorActionPreference = 'Stop'
[Console]::OutputEncoding = [System.Text.Encoding]::UTF8

$projectRoot = $PSScriptRoot
$router = Join-Path $projectRoot 'router.php'
if (-not (Test-Path $router)) {
    Write-Host 'ERROR: router.php not found in project root.' -ForegroundColor Red
    exit 1
}

function Find-PhpExecutable {
    $candidates = @(
        'php.exe',
        'C:\\php\\php.exe',
        'C:\\xampp\\php\\php.exe',
        'C:\\Program Files\\PHP\\php.exe'
    )
    foreach ($candidate in $candidates) {
        $cmd = Get-Command $candidate -ErrorAction SilentlyContinue
        if ($cmd) { return $cmd.Source }
        if (Test-Path $candidate) { return $candidate }
    }
    return $null
}

$php = Find-PhpExecutable
if (-not $php) {
    Write-Host 'ERROR: php.exe not found. Install PHP and add it to PATH.' -ForegroundColor Red
    exit 1
}

$env:APP_URL = 'http://localhost:8080'
$env:APP_DEBUG = 'true'
$env:SESSION_SECURE = 'false'
$env:DB_HOST = '127.0.0.1'
$env:DB_PORT = '3307'
$env:DB_NAME = 'furniture_platform'
$env:DB_USER = 'root'
if (-not $env:DB_PASSWORD) { $env:DB_PASSWORD = '' }

Write-Host 'Starting development server at http://localhost:8080' -ForegroundColor Cyan
Set-Location $projectRoot
& $php -S 'localhost:8080' -t $projectRoot $router
'''
script.write_text(contents, encoding='utf-16')
print('Wrote new start-dev.ps1', script)
```
